## Top TF-IDF Terms by Intent

To identify the most representative words for each intent, TF-IDF was applied to the preprocessed text. Unlike raw word frequency, TF-IDF highlights words that are highly informative for a specific intent while reducing the importance of common words shared across multiple intents.

A representative sample of labels is presented below for readability.
